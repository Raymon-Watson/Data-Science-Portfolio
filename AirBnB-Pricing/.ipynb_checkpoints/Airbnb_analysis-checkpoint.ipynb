# Airbnb Analysis
The aim of this project is to predict AirBnB rental prices based on listed features using a Linear Regression model.

## 1. Load and Inspect Data

````python
# Core Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Preprocessing and Statistical Methods
from sklearn.preprocessing import StandardScaler, LabelEncoder, PolynomialFeatures, MinMaxScaler, RobustScaler
from sklearn.model_selection import train_test_split, GridSearchCV, cross_val_score, cross_val_predict, cross_validate, StratifiedKFold
from sklearn.metrics import (
    r2_score
)

# Models
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor

from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.ensemble import HistGradientBoostingRegressor
````

````python
df = pd.read_csv("data/airbnb_data.csv") # training data
df.head()
````

## 2. Data Preparation

As there is not much preparation involved, we do it first before inspecting the distributions in detail.

## 2.1 Label Encoding

It is likely that room type, neighbourhood, and neighbourhood_group will be correlated with price. Here we encode these into numerical values.

````python
label_encoder = LabelEncoder()

df["room_type"] = label_encoder.fit_transform(df["room_type"])
df["neighbourhood"] = label_encoder.fit_transform(df["neighbourhood"])
df["neighbourhood_group"] = label_encoder.fit_transform(df["neighbourhood_group"])

df.head()
````

### 2.2 Drop Columns

````python
cols = ["latitude",
        "longitude",
        "price",
        "minimum_nights",
        "number_of_reviews",
        "reviews_per_month",
        "calculated_host_listings_count",
        "availability_365",
        "room_type",
        "neighbourhood",
        "neighbourhood_group"]

df = df[cols]
df.head()
````

### 2.3 Impute Data (reviews_per_month)

````python
df["reviews_per_month"]= df["reviews_per_month"].fillna(0)
print(df[cols].isnull().sum())
````

### 2.4 Train/Test Split

````python
X = df.drop("price", axis=1)
y = df["price"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=50)
````

All clean and ready to Linearly Regress!

## 4. Model the Data

### 4.1 Baseline Model (Linear Regression)
First, we attempt a simple Linear Regression model to check how well it can fit the training data. This is unlikely to do well since, as we noticed in the EDA stage, there are a number of outliers present in various features, additionally many of the features are heavily skewed.

````python
lin_reg = LinearRegression()
lin_crossval = cross_val_score(lin_reg, X_train, y_train, cv=5)
mean_lin_crossval = np.mean(lin_crossval)
std_lin_crossval = np.std(lin_crossval)
print(f"5-fold cross-validation for linear model: {np.round(mean_lin_crossval,3)} ± {np.round(std_lin_crossval,3)}")
````

As expected, we obtain a very low R2 score for the baseline model.

### 4.2 Tweaking Baseline Model
Before moving on to testing other models, we can perform tweaks of the baseline Linear Regression model to see if we can improve this result. To do this, we attempt the following methods:
- Including polynomial features
- Transforming skewed variables
- Regularized Linear Regression

#### 4..a Polynomial Features

````python
for deg in range(2,5):
    poly = PolynomialFeatures(degree=deg, include_bias=False)
    X_poly = poly.fit_transform(X_train)
    lin_crossval = cross_val_score(lin_reg, X_poly, y_train, cv=5, scoring="r2")
    mean_lin_crossval = np.mean(lin_crossval)
    std_lin_crossval = np.std(lin_crossval)
    print(f"5-fold cross-validation with {deg} polynomial features: {np.round(mean_lin_crossval,3)} ± {np.round(std_lin_crossval,3)}")
````

It seems including polynomial features of degree 2 improves the model fit negligibly, and higher order features radically degrade the fit. This means it is unlikely that this method can be used to improve the baseline model fit.

#### 4.2.b Transforming Skewed Variables
Recall that the heavily skewed columns we identified were:
- minimum_nights
- number_of_reviews
- reviews_per_month
- calculated_host_listings_count

**Logarithmic scaling:**

````python
X_scaled = X_train.copy()

num_cols = ["minimum_nights",
        "number_of_reviews",
        "reviews_per_month",
        "calculated_host_listings_count"]

for col in num_cols:
    X_scaled[f"{col}_log"] = np.log1p(X_scaled[col])


lin_crossval = cross_val_score(lin_reg, X_scaled, y_train, cv=5)
mean_lin_crossval = np.mean(lin_crossval)
std_lin_crossval = np.std(lin_crossval)
print(f"5-fold cross-validation for linear model: {np.round(mean_lin_crossval,3)} ± {np.round(std_lin_crossval,3)}")
````

**Square-root scaling:**

````python
X_scaled = X_train.copy()

num_cols = ["minimum_nights",
        "number_of_reviews",
        "reviews_per_month",
        "calculated_host_listings_count"]

for col in num_cols:
    X_scaled[f"{col}_sqrt"] = np.sqrt(X_scaled[col])


lin_crossval = cross_val_score(lin_reg, X_scaled, y_train, cv=5)
mean_lin_crossval = np.mean(lin_crossval)
std_lin_crossval = np.std(lin_crossval)
print(f"5-fold cross-validation for linear model: {np.round(mean_lin_crossval,3)} ± {np.round(std_lin_crossval,3)}")
````

**Cube-root scaling:**

````python
X_scaled = X_train.copy()

num_cols = ["minimum_nights",
        "number_of_reviews",
        "reviews_per_month",
        "calculated_host_listings_count"]

for col in num_cols:
    X_scaled[f"{col}_cbrt"] = np.cbrt(X_scaled[col])


lin_crossval = cross_val_score(lin_reg, X_scaled, y_train, cv=5)
mean_lin_crossval = np.mean(lin_crossval)
std_lin_crossval = np.std(lin_crossval)
print(f"5-fold cross-validation for linear model: {np.round(mean_lin_crossval,3)} ± {np.round(std_lin_crossval,3)}")
````

**Robust scaling:** Note that this is on *all features*

````python
rob_scal = RobustScaler()
X_scaled = rob_scal.fit_transform(X_train)

lin_crossval = cross_val_score(lin_reg, X_scaled, y_train, cv=5)
mean_lin_crossval = np.mean(lin_crossval)
std_lin_crossval = np.std(lin_crossval)
print(f"5-fold cross-validation for linear model: {np.round(mean_lin_crossval,3)} ± {np.round(std_lin_crossval,3)}")
````

Interestingly, the cube-root scaler improved the cross-validation scores the most, even if it is still only minor.

This suggests that a simple linear model is perhaps not the best method for fitting this data set. So we move on to other models to compare against our poor baseline.

#### 4.2.c Regularization
Before attempting these methods, let's make sure our features are **standardized**, since regression methods are sensitive to feature scale.

````python
std_scal = StandardScaler()
X_scaled = std_scal.fit_transform(X_train)
````

**Ridge Regression:**

````python
from sklearn.linear_model import Ridge

ridge_reg = Ridge(alpha=0.1)

scores = cross_val_score(ridge_reg, X_scaled, y_train, cv=5, scoring='r2')
print(f"Ridge CV R2: {np.round(scores.mean(),3)} ± {np.round(scores.std(),3)}")
````

**Lasso Regression:**

````python
from sklearn.linear_model import Lasso

lasso_reg = Lasso(alpha=0.1, max_iter=1000)

scores = cross_val_score(lasso_reg, X_scaled, y_train, cv=5, scoring='r2')
print(f"Ridge CV R2: {np.round(scores.mean(),3)} ± {np.round(scores.std(),3)}")
````

**Observations:** None of the methods to alter Linear Regression have improved upon the baseline model. It is unlikely that tuning the parameters much further will improve performance, so we move on to alternate models.

### 4.3 Train Alternative Models
The models we try in this section:
- Decision Tree Regressor
- Random Forest Regressor
- K-Nearest Neighbors
- SVR

#### 4.3.a Decision Tree Regressor

````python
from sklearn.tree import DecisionTreeRegressor

tree_reg = DecisionTreeRegressor(random_state=50)

scores = cross_val_score(tree_reg, X_train, y_train, cv=5, scoring="r2")

print(f"Decision Tree CV R2: {np.round(scores.mean(),3)} ± {np.round(scores.std(),3)}")
````

This is obviously incredibly bad. Let's do a quick check to see whether the max depth affects the results:

````python
for depth in range(1,10):
    tree_reg = DecisionTreeRegressor(random_state=50, max_depth=depth)
    scores = cross_val_score(tree_reg, X_train, y_train, cv=5, scoring="r2")
    print(f"Decision Tree CV R2 depth {depth}: {np.round(scores.mean(),3)} ± {np.round(scores.std(),3)}")
````

Unrestricted tree clearly performs quite badly, and limiting the max depth improves the result, but not by a great margin.

#### 4.3.b Random Forest Regressor

````python
from sklearn.ensemble import RandomForestRegressor

rf_reg = RandomForestRegressor(n_estimators=100, random_state=50)

scores = cross_val_score(rf_reg, X_train, y_train, cv=5, scoring="r2")

print(f"Random Forest CV R2: {np.round(scores.mean(),3)} ± {np.round(scores.std(),3)}")
````

**Best model so far**, and without any tuning.

#### 4.3.c K-Nearest Neighbors
Remember that this model is very sensitive to distance, so we need to use the standardized X features.

````python
from sklearn.neighbors import KNeighborsRegressor

knn_reg = KNeighborsRegressor(n_neighbors=5)

scores = cross_val_score(knn_reg, X_scaled, y_train, cv=5, scoring="r2")

print(f"KNN CV R2: {np.round(scores.mean(),3)} ± {np.round(scores.std(),3)}")
````

Really bad. Let's check the dependence on n_neighbors just in case:

````python
k_vals = range(1,30)
KNN_scores = []
for k in k_vals:
    knn_reg = KNeighborsRegressor(n_neighbors=k)

    scores = cross_val_score(knn_reg, X_scaled, y_train, cv=5, scoring="r2")
    KNN_scores.append(np.round(scores.mean(),3))


plt.plot(k_vals,KNN_scores)
plt.xlabel("k")
plt.ylabel("R2 Scores")
plt.title("k-Nearest Neighbors")
#plt.ylim(0,0.2)

````

Plateaus below the most basic Random Forest Regressor score, not worth pursuing.

#### 4.3.d SVR
Also requires standardized data.

````python
from sklearn.svm import SVR

SVR_reg = SVR()

scores = cross_val_score(SVR_reg, X_scaled, y_train, cv=5, scoring="r2")

print(f"SVR CV R2: {np.round(scores.mean(),3)} ± {np.round(scores.std(),3)}")
````

Again quite bad. Let's just check whether changing C alters the results:

````python
C_vals = [0.1, 1, 10, 100]
SVR_scores = []
for C in C_vals:
    SVR_reg = SVR(C=C)

    scores = cross_val_score(SVR_reg, X_scaled, y_train, cv=5, scoring="r2")
    SVR_scores.append(np.round(scores.mean(),3))


plt.plot(C_vals,SVR_scores)
plt.xlabel("C")
plt.ylabel("R2 Scores")
plt.title("SVR vs C")

````

**Observations:** Random Forest Regressor was the best performing without any tweaks. So we take this model to the next section so that we can potentially fine-tune it to improve the R2 score.

## 5. Fine-Tuning Random Forest
We begin by performing GridSearchCV:

````python
rf_reg = RandomForestRegressor( random_state=50)

param_grid = {
    "n_estimators": [100, 300, 500],
    "max_depth": [5, 10, 20],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4],
    "max_features": ["sqrt", 0.5, 1.0]
}

grid_search = GridSearchCV(estimator=rf_reg, param_grid=param_grid, cv=3, scoring="r2", verbose=True)

grid_search.fit(X_train, y_train)

print("Best parameters:")
print(grid_search.best_params_)

print("Best CV R2:")
print(grid_search.best_score_)
````

It looks like the best R2 is still quite low. This model is potentially **over-fitting**. Let's check the train R2 score to find out if this is the case:

````python
from sklearn.model_selection import cross_validate
best_rf = grid_search.best_estimator_

scores = cross_validate(
    best_rf,
    X_train,
    y_train,
    cv=5,
    scoring="r2",
    return_train_score=True
)

print("Train R2:", scores["train_score"].mean())
print("CV R2:", scores["test_score"].mean())
````

This is a clear sign that we are **over-fitting**, since the train R2 is very high, but the CV R2 is significantly lower.

We can attempt to improve this by reducing the complexity of our fit:
- Lower the maximum depth to decrease the model complexity
- Increasing the min_samples_leaf and min_samples_split

These all make the trees generated less complex, so potentially less prone to over-fitting.

````python
param_grid = {
    "n_estimators": [300, 500],
    "max_depth": [10, 15, 20, None],
    "min_samples_split": [2, 5, 10],
    "min_samples_leaf": [1, 2, 4, 6],
    "max_features": ["sqrt", 0.5, 0.8, 1.0]
}

grid_search = GridSearchCV(estimator=rf_reg, param_grid=param_grid, cv=5, scoring="r2", verbose=True)

grid_search.fit(X_train, y_train)

print("Best parameters:")
print(grid_search.best_params_)

print("Best CV R2:")
print(grid_search.best_score_)
````

````python
best_rf = grid_search.best_estimator_

scores = cross_validate(
    best_rf,
    X_train,
    y_train,
    cv=5,
    scoring="r2",
    return_train_score=True
)

print("Train R2:", scores["train_score"].mean())
print("CV R2:", scores["test_score"].mean())
````

By limiting the complexity of our model, we have very slightly increased the CV R2, but significantly reduced the Train R2.

Let's have a quick look at the feature importance to see if we can learn anything from this.

````python
importance = pd.Series(
    best_rf.feature_importances_,
    index=X_train.columns
).sort_values(ascending=False)

print(importance)
````

It is interesting that longitude and lattitude are the most important features. It could be worthwhile to generate a new feature that represents the distance from the city centre? Or some other feature that utilizes longitude and latitude.

## 7. Additional Feature Engineering
At this point, it seems unlikely that further tuning the model will improve our CV R2 score much. Instead, we can use what we have learned so far, in particular from the feature importance from the Random Forest model, to engineer potentially useful features.

Additionally, we can utilize logarithmic or cubic scaling for heavily skewed features.

We note that the Manhattan Center building has coordinates: 40.7527° N, 73.9943° W.

````python
X_eng = X_train.copy()

num_cols = ["minimum_nights",
        "number_of_reviews",
        "reviews_per_month",
        "calculated_host_listings_count"]

for col in num_cols:
    X_eng[f"{col}_log"] = np.log1p(X_eng[col])

# fraction of days available
X_eng["availability_fraction"] = X_eng["availability_365"] / 365

# reviews per available day
X_eng["reviews_per_available_day"] = X_eng["number_of_reviews"] / (X_eng["availability_365"] + 1)
# Added 1 to avoid dividing by 0


# distance from city centre
centre_longitude = -73.9943
centre_latitude = 40.7527
X_eng["distance_to_centre"] = np.sqrt( (X_eng["longitude"]-centre_longitude)**2 
                                    + (X_eng["latitude"]-centre_latitude)**2 )
````

````python
scores = cross_val_score(rf_reg, X_eng, y_train, cv=5, scoring="r2")

print(f"Random Forest CV R2: {np.round(scores.mean(),3)} ± {np.round(scores.std(),3)}")
````

````python
rf_reg.fit(X_eng, y_train)


importance = pd.Series(
    rf_reg.feature_importances_,
    index=X_eng.columns
).sort_values(ascending=False)

print(importance)
````

An extremely minor improvement to the fit.

At this point, it would be worthwhile to transform the target variable to see whether this is an important factor to our model fitting.

## 8. Transform Target Variable
We noted in the EDA section that the target variable was also heavily skewed. So let's use a logarithmic transform on the base data set (i.e. without any transformations on X_train) to check whether this is a dominant factor in our fit quality.

````python
y_log = np.log1p(y_train)
scores = cross_validate(
    rf_reg,
    X_train,
    y_log,
    cv=5,
    scoring="r2",
    return_train_score=True
)

print("Train CV R²:", scores["train_score"].mean())
print("Validation CV R²:", scores["test_score"].mean())
````

This represents a huge improvement in both training and validation R2 results! It is still over-fitting though, so we can focus on improving this:

````python
rf_reg = RandomForestRegressor(
    n_estimators=500,
    max_depth=15,
    min_samples_split=5,
    min_samples_leaf=2,
    max_features=0.8,
    random_state=42,
    n_jobs=-1
)

scores = cross_validate(
    rf_reg,
    X_train,
    y_log,
    cv=5,
    scoring="r2",
    return_train_score=True
)

print("Train CV R²:", scores["train_score"].mean())
print("Validation CV R²:", scores["test_score"].mean())
````

Massively improved train and validation R2 scores, though we are still minorly over-fitting. Let's move on to fine-tuning the model.

## 9. Fine-Tuning Final Model (Random Forest, log price)

````python
rf_reg = RandomForestRegressor( random_state=50)

param_grid = {
    "n_estimators": [100, 300, 500],
    "max_depth": [5, 10, 15],
    "min_samples_split": [5, 10],
    "min_samples_leaf": [3, 4, 5],
    "max_features": ["sqrt", 0.5, 1.0]
}

grid_search = GridSearchCV(estimator=rf_reg, param_grid=param_grid, cv=5, scoring="r2", verbose=True)

grid_search.fit(X_train, y_log)

print("Best parameters:")
print(grid_search.best_params_)

print("Best CV R2:")
print(grid_search.best_score_)

best_rf = grid_search.best_estimator_

scores = cross_validate(
    best_rf,
    X_train,
    y_log,
    cv=5,
    scoring="r2",
    return_train_score=True
)

print("Train R2:", scores["train_score"].mean())
print("CV R2:", scores["test_score"].mean())
````

A very minor improvement to the CV R2, and a reduction to the Train R2, which suggests we have lowered overfitting minorly. Further tuning could be useful, though it would likely be more useful to prune outliers from our data. 

As such, we move forward to evaluating this model on the test set.

## 10. Final Results

````python
from sklearn.metrics import r2_score

# Convert actual prices to log scale
y_test_log = np.log1p(y_test)

# Predictions from model trained on log(price)
y_pred_log = best_rf.predict(X_test)

log_r2_score = r2_score(y_test_log, y_pred_log)
y_pred = np.expm1(y_pred_log)
r2_score = r2_score(y_test, y_pred)


plt.figure(figsize=(7, 6))

plt.scatter(
    y_test_log,
    y_pred_log,
    alpha=0.4
)

# Perfect prediction line
low = min(y_test_log.min(), y_pred_log.min())
high = max(y_test_log.max(), y_pred_log.max())

plt.plot(
    [low, high],
    [low, high],
    "--"
)

plt.xlabel("Actual log(1 + Price)")
plt.ylabel("Predicted log(1 + Price)")
plt.title("Actual vs Predicted Log Price")

plt.show()

print(f"R2 log: {log_r2_score}")
````

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 6))

plt.scatter(y_test, y_pred, alpha=0.4)

# Perfect-prediction line
low = min(y_test.min(), y_pred.min())
high = max(y_test.max(), y_pred.max())

plt.plot([low, high], [low, high], "--")

plt.xlabel("Actual Price")
plt.ylabel("Predicted Price")
plt.title("Actual vs Predicted Airbnb Price")

plt.show()

print(f"R2: {r2_score}")
````

````python
from sklearn.metrics import median_absolute_error

print("Original-scale MAE:", mean_absolute_error(y_test, y_pred))

print("Median absolute error:", median_absolute_error(y_test, y_pred))
````

The model captures the overall relationship in log-price reasonably well, but predictions are compressed toward the centre of the distribution. In particular, the most expensive listings tend to be underpredicted, indicating that the model has difficulty capturing extreme prices.
